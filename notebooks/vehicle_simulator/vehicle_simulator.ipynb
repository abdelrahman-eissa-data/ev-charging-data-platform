# charging session simulieren 
## 1-Random Werte Erstellung

```python
import random
from datetime import datetime, timedelta

battery_start = random.randint(10, 40)
battery_end = random.randint(60, 95)

print(f"Battery Start: {battery_start}%")
print(f"Battery End:   {battery_end}%")
```

## 2-Eine komplette Session

```python
import random
from datetime import datetime, timedelta

battery_capacity = 75  # kWh — Tesla Model 

battery_start = random.randint(10, 40)
battery_end   = random.randint(battery_start + 20, 95)
energy_kwh = (battery_capacity * (battery_end - battery_start)) / 100

print(f"Battery Start: {battery_start}%")
print(f"Battery End:   {battery_end}%")
print(f"Energy:        {energy_kwh} kWh")
```

## 3-Ladedauer berechnen (in Minuten)

```python
power_kw = 50  # Ladegeschwindigkeit
duration_hours = energy_kwh / power_kw
duration_minutes = duration_hours * 60

print(f"Duration: {round(duration_minutes)} Minuten")
```

## 4-Start Zeit

```python
start_time = datetime.now().replace(microsecond=0)
end_time = start_time + timedelta(minutes=duration_minutes)

print(f"Start: {start_time}")
print(f"End:   {end_time}")
```

## 5- Eine Session als Dictionary

```python
import json

session = {
    "vehicle_id":        "EV_001",
    "battery_start_pct": battery_start,
    "battery_end_pct":   battery_end,
    "energy_kwh":        energy_kwh,
    "duration_minutes":  round(duration_minutes),
    "start_time":        start_time,
    "end_time":          end_time,
    "power_kw":          power_kw
}

print(json.dumps(session, indent=2, default=str)) # default=str → weil datetime kein JSON Format ist
```

## 6- 100 Sessions simulieren

```python
sessions = []

for i in range(100):
   
    battery_start = random.randint(10, 40)
    battery_end = random.randint(battery_start + 20, 95)
    energy_kwh = (75 * (battery_end - battery_start)) / 100
    duration_minutes = round((energy_kwh / 50) * 60)
    start_time = datetime.now().replace(microsecond=0)
    end_time = start_time + timedelta(minutes=duration_minutes)

    session = {
        "vehicle_id": f"EV_{i+1:03d}",  # EV_001, EV_002...
        "battery_start_pct": battery_start,
        "battery_end_pct": battery_end,
        "energy_kwh": energy_kwh,
        "duration_minutes": duration_minutes,
        "start_time": start_time,
        "end_time": end_time,
        "power_kw": 50
    }

    sessions.append(session)

print(f"Total sessions: {len(sessions)}")
print(sessions[0])
print(sessions[1])
```

## 7- DataFrame erstellen

```python
import pandas as pd 

df = pd.DataFrame(sessions)

display(df.head())

df.info()

#alle Start datum sind gleich 
```
